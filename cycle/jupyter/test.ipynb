```python
import re

input1 = "{1,2,3},{4,5,6}"
input2 = "{1,2,3},2"
input3 = "[1,2,3],[4,5,6]"
input4 = "7,[[1,2,8,9],[2,4,9,12],[4,7,10,13],[6,8,11,15]]"
```

```python
list_match1 = re.findall(r'\{([^{}]*)\}', input1)
print(list_match1)
num_list1 = list(map(int,re.findall(r'-?\d+',list_match1[0])))
print(num_list1)
```

```python
list_match2 = re.search(r'\{([^{}]*)\}', input2)
print(list_match2.group(1))
num_list2 = list_match2.group(1)
num_list_2 = list(map(int,re.findall(r'-?\d+',num_list2)))
print(num_list_2)
value_match = re.sub(r'\{([^{}]*)\}',"",input2)
value_list = list(map(int,re.findall(r'-?\d+',value_match)))
value = value_list[0]
print(value)
```

```python
list_match3 = re.findall(r'\[([^\]]*)\]', input3)
print(list_match3)
values_list3_1 = list(map(int,re.findall(r'=?\d+',list_match3[0])))
values_list3_2 = list(map(int,re.findall(r'=?\d+',list_match3[1])))
print(values_list3_1)
```

```python
value_match = re.findall(r'-?\d+',input4)
value = int(value_match[0])
print(value)
list_match = re.findall(r'\[([^\]]*)\]', input4)
res = []
for num_list in list_match:
    num_list = list(map(int,re.findall(r'-?\d+',num_list)))
    res.append(num_list)
print(res)
```
